```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

df = pd.read_csv("../data/raw/ai4i2020.csv")

feature_columns = [
    "Type",
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

X = df[feature_columns].copy()
y = df["Machine failure"].copy()

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

numeric_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

categorical_features = ["Type"]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Preparation complete!")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.pipeline import Pipeline

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=8,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=None,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}

trained_models = {}

for name, model in models.items():

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("classifier", model)
        ]
    )

    pipeline.fit(X_train, y_train)

    trained_models[name] = pipeline

    print(f"{name} trained successfully!")

print("\nAll baseline models trained!")
```

```python
predictions = {}
probabilities = {}

for name, pipeline in trained_models.items():

    predictions[name] = pipeline.predict(X_test)

    probabilities[name] = pipeline.predict_proba(X_test)[:, 1]

print("Predictions generated successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

results = []

for name in trained_models:

    y_pred = predictions[name]
    y_prob = probabilities[name]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, y_pred, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test, y_prob),
        "PR-AUC (AP)": average_precision_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="PR-AUC (AP)",
    ascending=False
)

results_df.round(4)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import confusion_matrix

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

for ax, (name, y_pred) in zip(
    axes,
    predictions.items()
):

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["No Failure", "Failure"],
        yticklabels=["No Failure", "Failure"],
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import classification_report

for name, y_pred in predictions.items():

    print("=" * 60)
    print(name)
    print("=" * 60)

    print(
        classification_report(
            y_test,
            y_pred,
            target_names=["No Failure", "Failure"],
            zero_division=0
        )
    )
```

```python
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

from sklearn.pipeline import Pipeline

# Calculate imbalance ratio from training data only
negative_count = int((y_train == 0).sum())
positive_count = int((y_train == 1).sum())

scale_pos_weight = negative_count / positive_count

advanced_models = {
    "XGBoost": XGBClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.9,
        colsample_bytree=0.9,
        scale_pos_weight=scale_pos_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        tree_method="hist",
        random_state=42,
        n_jobs=-1
    ),

    "LightGBM": LGBMClassifier(
        n_estimators=300,
        learning_rate=0.05,
        num_leaves=31,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1,
        verbosity=-1
    ),

    "CatBoost": CatBoostClassifier(
        iterations=300,
        depth=6,
        learning_rate=0.05,
        loss_function="Logloss",
        eval_metric="AUC",
        auto_class_weights="Balanced",
        random_seed=42,
        verbose=0,
        allow_writing_files=False,
        thread_count=-1
    )
}

trained_advanced_models = {}

for name, model in advanced_models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", model)
    ])

    pipeline.fit(X_train, y_train)

    trained_advanced_models[name] = pipeline

    print(f"{name} trained successfully!")

print("\nAll advanced models trained!")
```

```python
advanced_predictions = {}
advanced_probabilities = {}

for name, pipeline in trained_advanced_models.items():

    advanced_predictions[name] = pipeline.predict(X_test)

    advanced_probabilities[name] = (
        pipeline.predict_proba(X_test)[:, 1]
    )

print("Advanced model predictions generated!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

advanced_results = []

for name in trained_advanced_models:

    y_pred = advanced_predictions[name]
    y_prob = advanced_probabilities[name]

    advanced_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, y_pred, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test, y_prob),
        "PR-AUC (AP)": average_precision_score(y_test, y_prob)
    })

advanced_results_df = pd.DataFrame(advanced_results)

advanced_results_df = advanced_results_df.sort_values(
    by="PR-AUC (AP)",
    ascending=False
)

advanced_results_df.round(4)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import confusion_matrix

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

for ax, (name, y_pred) in zip(
    axes,
    advanced_predictions.items()
):

    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["No Failure", "Failure"],
        yticklabels=["No Failure", "Failure"],
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

plt.tight_layout()
plt.show()
```

```python
# STEP 13.2: Hyperparameter Tuning using RandomizedSearchCV

import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# Build the pipeline
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(random_state=42, n_jobs=1))
])

# Parameters to explore
param_distributions = {
    "classifier__n_estimators": [100, 200, 300, 400],
    "classifier__max_depth": [None, 10, 15, 20, 25],
    "classifier__min_samples_split": [2, 5, 10],
    "classifier__min_samples_leaf": [1, 2, 4],
    "classifier__max_features": ["sqrt", 0.8],
    "classifier__class_weight": ["balanced", "balanced_subsample"]
}

# Cross-validation strategy
cv_strategy = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

# Search for better hyperparameters
rf_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_distributions,
    n_iter=12,
    scoring="average_precision",
    cv=cv_strategy,
    random_state=42,
    n_jobs=-1,
    verbose=1,
    refit=True
)

# Train using training data only
rf_search.fit(X_train, y_train)

# Display the best settings
print("\nBest Hyperparameters:")
print(rf_search.best_params_)

print("\nBest Cross-Validation PR-AUC:")
print(round(rf_search.best_score_, 4))

print("\nTuning completed successfully!")
```

```python
# Evaluate the tuned Random Forest on the test set

best_rf_model = rf_search.best_estimator_

y_pred_tuned = best_rf_model.predict(X_test)
y_prob_tuned = best_rf_model.predict_proba(X_test)[:, 1]

tuned_results = pd.DataFrame([{
    "Model": "Tuned Random Forest",
    "Accuracy": accuracy_score(y_test, y_pred_tuned),
    "Precision": precision_score(
        y_test, y_pred_tuned, zero_division=0
    ),
    "Recall": recall_score(
        y_test, y_pred_tuned, zero_division=0
    ),
    "F1 Score": f1_score(
        y_test, y_pred_tuned, zero_division=0
    ),
    "ROC-AUC": roc_auc_score(y_test, y_prob_tuned),
    "PR-AUC (AP)": average_precision_score(
        y_test, y_prob_tuned
    )
}])

display(tuned_results.round(4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_tuned,
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_tuned))
```

```python
# STEP 13.3.1: Compare baseline, advanced and tuned models

# Combine previously evaluated models
comparison_parts = []

if "results_df" in globals():
    comparison_parts.append(results_df.copy())

if "advanced_results_df" in globals():
    comparison_parts.append(advanced_results_df.copy())

if "tuned_results" in globals():
    comparison_parts.append(tuned_results.copy())

if not comparison_parts:
    raise NameError(
        "Model results not found. Run STEP 11 and STEP 12 evaluation cells first."
    )

final_comparison = pd.concat(
    comparison_parts, ignore_index=True
)

# Keep only the best row if a model name appears more than once
final_comparison = (
    final_comparison
    .sort_values("PR-AUC (AP)", ascending=False)
    .drop_duplicates(subset="Model", keep="first")
    .reset_index(drop=True)
)

print("FINAL MODEL COMPARISON")
display(final_comparison.round(4))
```

```python
# Rank models using PR-AUC, then Recall

final_comparison = final_comparison.sort_values(
    by=["PR-AUC (AP)", "Recall", "F1 Score"],
    ascending=False
).reset_index(drop=True)

print("Model ranking:")
display(final_comparison.round(4))

selected_model_name = final_comparison.iloc[0]["Model"]

print("\nSelected model based on PR-AUC:")
print(selected_model_name)

# Link the selected name to the trained pipeline
if selected_model_name == "Tuned Random Forest":
    final_model = best_rf_model

elif selected_model_name in globals().get("trained_models", {}):
    final_model = trained_models[selected_model_name]

elif selected_model_name in globals().get("trained_advanced_models", {}):
    final_model = trained_advanced_models[selected_model_name]

else:
    raise ValueError(
        f"Trained pipeline not found for: {selected_model_name}"
    )

print("\nFinal model pipeline is ready!")
```

```python
# STEP 17.1: Extract feature importance

import pandas as pd
import matplotlib.pyplot as plt

# Use the selected final model
classifier = final_model.named_steps["classifier"]
preprocessor_fitted = final_model.named_steps["preprocessor"]

# Obtain the transformed feature names
feature_names = preprocessor_fitted.get_feature_names_out()

# Extract importance from supported tree-based models
if hasattr(classifier, "feature_importances_"):
    importance_values = classifier.feature_importances_

elif hasattr(classifier, "coef_"):
    importance_values = abs(classifier.coef_[0])

else:
    raise ValueError(
        "This model does not expose built-in feature importance. "
        "We can use permutation importance instead."
    )

feature_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance_values
}).sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

display(feature_importance_df.round(4))

# Plot the top features
top_features = feature_importance_df.head(10).sort_values(
    by="Importance",
    ascending=True
)

plt.figure(figsize=(10, 6))
plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Feature Importance — Final Model")
plt.tight_layout()
plt.show()
```
